Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и `*args`. Создайте класс `DocumentDesk` с методом `issue(pupil, reason=None, teacher=None)`, который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлена справка ученику: <name>», «Оформлена справка ученику <name> по причине: <reason>» или «Оформлена справка ученику <name> по причине <reason>, заверил: <teacher>».
```python
class DocumentDesk:
    def issue(self, pupil, reason=None, teacher=None):
        if reason is None:
            print(f"Оформлена справка ученику: {pupil}")
        elif teacher is None:
            print(f"Оформлена справка ученику {pupil} по причине: {reason}")
        else:
            print(f"Оформлена справка ученику {pupil} по причине {reason}, заверил: {teacher}")

d = DocumentDesk()
d.issue("Иван")
d.issue("Иван", "болезнь")
d.issue("Иван", "болезнь", "Петров П.П.")

# В Python имитируют перегрузку через значения по умолчанию и *args.
```

Дополнительно реализуйте метод `__repr__`, который возвращает техническое представление объекта, например `Pupil('Иван Петров', '9А', 4.5)` — оно используется при выводе списка учеников и в отладчике. Поясните разницу между `__str__` и `__repr__` и когда какой используется: `__str__` — для человека, `__repr__` — для разработчика.

Нужно, чтобы программа понимала, что это «один и тот же ученик», и не дублировала его в списках. Реализуйте в классе `Pupil` магический метод `__eq__`, чтобы два ученика считались равными, если совпадают их имя и класс. Дополнительно реализуйте `__hash__`, чтобы учеников можно было хранить во множестве `set` и использовать как ключи в словаре. Проверьте, что два одинаковых ученика попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении `__eq__` нужно также переопределять `__hash__` — что произойдёт, если этого не сделать.


```python
class Pupil:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def __eq__(self, other):
        if not isinstance(other, Pupil):
            return NotImplemented
        return self.name == other.name and self.grade == other.grade

    def __hash__(self):
        return hash((self.name, self.grade))

a = Pupil("Иван", "9А")
b = Pupil("Иван", "9А")
c = Pupil("Мария", "9А")

print(a == b)

s = {a, b, c}
print(len(s))  # 2

# Если переопределить __eq__, нужно переопределить и __hash__,
# иначе объекты станут нехешируемыми или словарь/множество будут работать неправильно.
```

**Задача 13 (магические методы `__add__` и `__len__`).** В школе есть параллели, и иногда нужно объединять их (например, при слиянии классов) или узнавать, сколько учеников в параллели. Создайте класс `Grade` (параллель), в котором хранится список учеников. Реализуйте магический метод `__add__` для объединения двух параллелей в одну (возвращается **новая** параллель со всеми учениками) и метод `__len__`, возвращающий количество учеников. Продемонстрируйте работу: создайте две параллели, объедините их через `g3 = g1 + g2` и выведите `len(g3)`. Объясните, почему `__add__` возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.


```python
class Grade:
    def __init__(self, pupils=None):
        self.pupils = list(pupils) if pupils else []

    def __add__(self, other):
        if not isinstance(other, Grade):
            return NotImplemented
        return Grade(self.pupils + other.pupils)

    def __len__(self):
        return len(self.pupils)

    def __repr__(self):
        return f"Grade({self.pupils!r})"

g1 = Grade(["Иван", "Мария"])
g2 = Grade(["Пётр"])
g3 = g1 + g2

print(len(g3))  # 3
print(g3)

# __add__ возвращает новый объект, чтобы не менять исходные параллели.
```

```python
class Pupil:
    def __init__(self, name, grade, average_mark):
        self.name = name
        self.grade = grade
        self.average_mark = average_mark

    @property
    def average_mark(self):
        return self._average_mark

    @average_mark.setter
    def average_mark(self, value):
        if not isinstance(value, (int, float)) or isinstance(value, bool):
            raise ValueError("Оценка должна быть числом")
        if not (0 <= value <= 5):
            raise ValueError("Оценка должна быть в диапазоне 0..5")
        self._average_mark = value

p = Pupil("Иван", "9А", 4.5)
print(p.average_mark)

for bad in (6, -1, "abc"):
    try:
        p.average_mark = bad
    except ValueError as e:
        print("ValueError:", e)

p.average_mark = 4.8
print(p.average_mark)

# "@property позволяет обращаться как к атрибуту, но выполнять проверку.
```

**Задача 15 (классовый метод `@classmethod`).** В школу часто поступают данные об учениках в виде строк — например, из файла или базы данных: `"Иван Петров;9А;4.5"`. Добавьте в класс `Pupil` классовый метод `from_string(cls, data)`, который создаёт объект ученика из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание ученика через этот метод и вывод результата. Объясните разницу между `@classmethod`, `@staticmethod` и обычным методом: обычный метод получает первым аргументом экземпляр (`self`) и работает с конкретным объектом, `@classmethod` получает сам класс (`cls`) и часто используется как альтернативный конструктор, `@staticmethod` не получает ни `self`, ни `cls` и представляет собой просто утилиту. Расскажите, когда какой использовать.


```python
class Pupil:
    def __init__(self, name, grade, average_mark):
        self.name = name
        self.grade = grade
        self.average_mark = average_mark

    @classmethod
    def from_string(cls, data):
        name, grade, mark = data.split(";")
        return cls(name, grade, float(mark))

    def __repr__(self):
        return f"Pupil({self.name!r}, {self.grade!r}, {self.average_mark!r})"

p = Pupil.from_string("Иван Петров;9А;4.5")
print(p)

# "@classmethod получает cls и часто используется как альтернативный конструктор.
# "@staticmethod не получает ни self, ни cls.
```

Создайте собственное исключение `LessonNotFoundError(Exception)` и используйте его в классе `School` (школа). Метод `enroll(pupil, lesson_name)` должен записывать ученика на урок, если он есть, и выбрасывать `LessonNotFoundError` с понятным сообщением (например, с указанием названия урока), если урока нет. Продемонстрируйте обработку исключения через `try/except` и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.


```python
class LessonNotFoundError(Exception):
    pass

class School:
    def __init__(self, lessons):
        self.lessons = set(lessons)

    def enroll(self, pupil, lesson_name):
        if lesson_name not in self.lessons:
            raise LessonNotFoundError(f"Урок '{lesson_name}' не найден в расписании")
        print(f"{pupil} записан на урок {lesson_name}")

school = School(["Математика", "Физика"])

try:
    school.enroll("Иван", "Химия")
except LessonNotFoundError as e:
    print("Поймано своё исключение:", e)

school.enroll("Иван", "Математика")

# Своё исключение помогает отличать ожидаемые ошибки школы от чужих ошибок программы.
```

**Задача 22 (дескрипторы).** Свойство `@property` — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор `RangeDescriptor(min_value, max_value)`, который хранит значение в `instance.__dict__` (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая `ValueError` при нарушении. Используйте этот дескриптор в классе `Pupil` для полей `average_mark` (например, диапазон 0–5) и `age` (диапазон 6–19), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» `@property`, зачем нужен метод `__set_name__` и в чём преимущество дескриптора перед `@property`, если одинаковая проверка повторяется в десяти полях.


```python
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __set_name__(self, owner, name):
        self.name = name

    def __get__(self, instance, owner):
        if instance is None:
            return self
        return instance.__dict__.get(self.name)

    def __set__(self, instance, value):
        if not isinstance(value, (int, float)) or isinstance(value, bool):
            raise ValueError(f"{self.name} должно быть числом")
        if not (self.min_value <= value <= self.max_value):
            raise ValueError(
                f"{self.name} должно быть в диапазоне {self.min_value}..{self.max_value}"
            )
        instance.__dict__[self.name] = value

class Pupil:
    average_mark = RangeDescriptor(0, 5)
    age = RangeDescriptor(6, 19)

    def __init__(self, name, average_mark, age):
        self.name = name
        self.average_mark = average_mark
        self.age = age

p = Pupil("Иван", 4.5, 15)
print(p.average_mark, p.age)

for field, value in [("average_mark", 6), ("age", 20)]:
    try:
        setattr(p, field, value)
    except ValueError as e:
        print("ValueError:", e)

# Дескриптор переиспользует логику для разных полей.
# "@property — это тоже дескриптор."
```

**Задача 23 (метаклассы).** В школе должен быть только один реестр учеников — если создать второй, данные могут разойтись. Создайте метакласс `SingletonMeta`, который гарантирует, что у класса `School` будет только один экземпляр. Проверьте, что `School()` дважды возвращает один и тот же объект (сравните через `is`). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.


```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class School(metaclass=SingletonMeta):
    pass

a = School()
b = School()
print(a is b)  # True

# Метакласс — класс класса.
# School — экземпляр SingletonMeta.
# Метаклассы редки, но важны для фреймворков.
```

**Задача 24 (паттерн Singleton).** Реализуйте класс `PupilRegistry` через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода `__new__` (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; `__new__` проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).


```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class PupilRegistryMeta(metaclass=SingletonMeta):
    pass

class PupilRegistryNew:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

r1 = PupilRegistryMeta()
r2 = PupilRegistryMeta()
print("metaclass:", r1 is r2)

r3 = PupilRegistryNew()
r4 = PupilRegistryNew()
print("__new__:", r3 is r4)

# Singleton удобен для кэша, конфига, логгера.
# Вреден, если мешает тестированию и скрывает зависимости.
```
