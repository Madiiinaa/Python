Задача 11: Определите класс LeaveRequest с полями имя сотрудника, начальная дата и конечная дата. Реализуйте методы update_dates(new_start_date, new_end_date), который обновляет даты отпуска, и display_info(), который выводит информацию о заявке.

```python
class LeaveRequest:
    def __init__(self, employee_name, start_date, end_date):
        self.employee_name = employee_name
        self.start_date = start_date
        self.end_date = end_date

    def update_dates(self, new_start_date, new_end_date):
        self.start_date = new_start_date
        self.end_date = new_end_date

    def display_info(self):
        print(f"Сотрудник: {self.employee_name}")
        print(f"Отпуск с {self.start_date} по {self.end_date}")


request = LeaveRequest("Рабаданова МР", "02.09.2026", "22.08.2026")
request.display_info()

print("\n Обновление дат ")
request.update_dates("15.09.2026", "29.09.2026")
request.display_info()
```

Задача 11. Определите класс LeaveRequest, который наследуется от класса Event. Класс LeaveRequest должен добавлять поля начальная дата и конечная дата, а также методы для обновления дат отпуска.

```python
class Event:
    def __init__(self, name):
        self.name = name

    def display_event(self):
        print(f"Событие: {self.name}")


class LeaveRequest(Event):
    def __init__(self, name, start_date, end_date):
        super().__init__(name)
        self.start_date = start_date
        self.end_date = end_date

    def update_dates(self, new_start_date, new_end_date):
        self.start_date = new_start_date
        self.end_date = new_end_date

    def display_info(self):
        self.display_event()
        print(f"Отпуск с {self.start_date} по {self.end_date}")


request = LeaveRequest("Отпуск сотрудника", "01.06.2026", "23.06.2026")
request.display_info()

print("\n Обновление дат ")
request.update_dates("05.07.2026", "21.07.2026")
request.display_info()
```

Задача 11. Создать класс LeaveRequest, у которого поля employeeName, startDate, endDate будут приватными. Предоставить методы для безопасного изменения дат отпуска и доступа к имени сотрудника.

```python
class LeaveRequest:
    def __init__(self, employee_name, start_date, end_date):
        self.__employee_name = employee_name
        self.__start_date = start_date
        self.__end_date = end_date

    def get_employee_name(self):
        return self.__employee_name

    def update_dates(self, new_start_date, new_end_date):
        if not new_start_date or not new_end_date:
            raise ValueError("Даты не могут быть пустыми")
        if new_start_date > new_end_date:
            raise ValueError("Начальная дата не может быть позже конечной")
        self.__start_date = new_start_date
        self.__end_date = new_end_date

    def get_start_date(self):
        return self.__start_date

    def get_end_date(self):
        return self.__end_date

    def display_info(self):
        print(f"Сотрудник: {self.__employee_name}")
        print(f"Отпуск с {self.__start_date} по {self.__end_date}")


request = LeaveRequest("Аитова ГР", "05.03.2026", "27.03.2026")
request.display_info()

print(f"\nИмя сотрудника: {request.get_employee_name()}")

print("\n Обновление дат ")
request.update_dates("06.04.2026", "25.04.2026")
request.display_info()

print("\n Попытка некорректного обновления ")
try:
    request.update_dates("31.09.2026", "01.09.2026")
except ValueError as e:
    print(f"Ошибка: {e}")
```
