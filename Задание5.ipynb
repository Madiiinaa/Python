Выведите его dtype, shape, size, ndim. Дополнительно создайте второй массив из тех же значений, но с типом float32, и сравните dtype и nbytes с первым. Ожидаемый вывод: для первого массива dtype=float64, shape=(7,), size=7, ndim=1; для второго — dtype=float32, nbytes=28. В комментариях поясните: почему NumPy выбрал именно float64 для первого массива; чем shape отличается от size; почему shape у одномерного массива — это кортеж из одного элемента (7,), а не (7); что произойдёт при обращении marks[7] и почему; как nbytes связан с itemsize и size.

Ожидаемый вывод: float64 — 7.63 МБ, float32 — 3.81 МБ, uint8 — 0.95 МБ. Дополнительно оцените размер Python-списка из 1 000 000 чисел через sys.getsizeof (только оболочки) и сравните с массивами NumPy. В комментариях объясните: почему nbytes не включает метаданные объекта (что ещё занимает память); почему для больших массивов выбор типа — это не микрооптимизация, а вопрос «поместится ли в RAM»; во сколько раз uint8 экономнее float64 и как это влияет на скорость обработки за счёт кэша процессора; в каких случаях экономия памяти важнее точности вычислений.

````python
import numpy as np, sys

n = 1_000_000
for dt in (np.float64, np.float32, np.uint8):
    a = np.zeros(n, dtype=dt)
    print(dt.__name__, a.itemsize, round(a.nbytes / 1024**2, 2), "МБ")

lst = [0.0] * n
print("list shell:", round(sys.getsizeof(lst)/1024**2, 2), "МБ")
print("list всего :", round((sys.getsizeof(lst) + n*sys.getsizeof(0.0))/1024**2, 2), "МБ")

# nbytes не учитывает заголовок ndarray (~100 байт) и Python-обёртку.
# uint8 в 8 раз экономнее float64 → влезает в кэш CPU → быстрее.
# Экономия важнее точности, когда оценки — целые (например, баллы ЕГЭ).
````

Задача 8 (случайные числа, воспроизводимость). Школа моделирует данные для симуляции успеваемости. Создайте генератор rng = np.random.default_rng(seed=42) и сгенерируйте: 10 равномерных значений rng.random(10) — вероятности сдать экзамен; 10 нормальных значений rng.normal(4.0, 0.5, 10) — средние оценки; 10 целых значений rng.integers(1, 12, 10) — номера классов. Создайте второй генератор с тем же seed=42 и убедитесь, что все три последовательности совпадают. Создайте третий генератор без seed и покажите, что значения отличаются. В комментариях объясните: зачем нужен seed и как он обеспечивает воспроизводимость экспериментов; чем default_rng отличается от старого np.random.seed (какой BitGenerator используется по умолчанию); почему в новом коде рекомендуется Generator, а не глобальный np.random; как SeedSequence помогает организовать независимые потоки случайных чисел для параллельных вычислений; в каких ситуациях отсутствие фиксированного seed — это хорошо, а в каких — плохо.

````python
import numpy as np

rng1 = np.random.default_rng(seed=42)
a1 = rng1.random(10)
b1 = rng1.normal(4.0, 0.5, 10)
c1 = rng1.integers(1, 12, 10)

rng2 = np.random.default_rng(seed=42)
a2 = rng2.random(10)
b2 = rng2.normal(4.0, 0.5, 10)
c2 = rng2.integers(1, 12, 10)

print("воспроизводимо:", np.array_equal(a1, a2),
      np.array_equal(b1, b2), np.array_equal(c1, c2))

rng3 = np.random.default_rng()  # без seed
print("без seed отличается:", not np.allclose(a1, rng3.random(10)))

# seed фиксирует состояние BitGenerator (PCG64 в default_rng).
# default_rng — новый API (Generator), рекомендован вместо np.random.seed.
# SeedSequence позволяет порождать независимые потоки (spawn).
# Без seed хорошо для криптографии/симуляций, плохо для воспроизводимости.
````

Выведите все три. Ожидаемый вывод для первой: диагональ из единиц, остальные нули. В комментариях объясните: чем np.eye(4) отличается от np.identity(4) (какой параметр позволяет задать вторую размерность); что делает параметр k в np.diag (положительное значение — диагональ выше главной, отрицательное — ниже); в каком случае диагональная матрица лучше обычной (память и скорость умножения, если большинство элементов нули); почему np.eye по умолчанию даёт float, а не int, и когда это важно.

В комментариях объясните: почему data - data.mean(axis=0) работает без явного reshape (как broadcasting растягивает (5,) до (365, 5)); что произойдёт, если перепутать axis=0 и axis=1 (какая ошибка или неверный результат); как broadcasting экономит память (не создаёт копию среднего на 365 строк); как то же самое сделать через keepdims=True и в каком случае это удобнее; зачем вообще нужна стандартизация перед анализом успеваемости.

В комментариях объясните: чем where отличается от select (когда два варианта, когда много); почему при перекрывающихся условиях select берёт первое (как это использовать для приоритетов); в каком случае clip лучше where (краткость, скорость, читаемость); чем reduce отличается от accumulate (свёртка в одно значение vs промежуточные результаты); что такое identity у ufunc (np.add.identity == 0) и зачем он нужен.

````python
import numpy as np

marks = np.array([2.9, 3.5, 4.2, 4.7, 5.0, 3.8, 2.5])

print(np.where(marks < 3.0, 3.0, marks))    # поднять минимум
print(np.where(marks > 5.0, 5.0, marks))    # обрезать максимум

conditions = [marks < 3.5,
              (marks >= 3.5) & (marks <= 4.5),
              marks > 4.5]
choices = ["троечник", "хорошист", "отличник"]
print(np.select(conditions, choices, default="?"))

print("clip:", np.clip(marks, 3.0, 5.0))

lessons_per_day = np.array([6, 5, 6, 5, 6, 4, 5])
print("reduce    :", np.add.reduce(lessons_per_day))       # 37
print("accumulate:", np.add.accumulate(lessons_per_day))   # [6 11 17 22 28 32 37]
print("identity  :", np.add.identity)                      # 0

# where — два варианта; select — много (первое подходящее условие побеждает).
# clip — кратче и быстрее, когда нужно обрезать диапазон.
# reduce → одно значение; accumulate → промежуточные результаты.
# identity — нейтральный элемент для reduce.
````

В комментариях объясните: почему partition работает за O(n), а sort — за O(n log n); что гарантирует partition (k-й элемент на своём месте, слева — меньше, справа — больше, но порядок внутри частей не определён); во сколько раз partition быстрее sort для 1 000 000 элементов (по экспериментальным данным); почему argpartition + argsort дают корректный топ-N (что нужно сделать после argpartition, чтобы упорядочить топ); в каком случае разница в скорости критична (интерактивные приложения, реальное время).

```
t = np.linspace(0, 1, 1000, endpoint=False)

noisy = signal + np.random.default_rng(42).normal(0, 0.3, 1000)

```

Выполните: np.fft.fft(noisy) — прямое преобразование; np.fft.fftfreq(len(noisy), d=1/1000) — частоты; amplitudes = np.abs(fft) / len(noisy) * 2 — амплитуды; найдите топ-2 частоты через np.argsort(amplitudes)[-2:]. Отфильтруйте шум: обнулите все частоты выше 20 Гц и выполните np.fft.ifft. Сравните ошибку до и после фильтрации. В комментариях объясните: что такое FFT (как раскладывает сигнал на частоты); почему длины, кратные степени двойки, эффективнее (какой алгоритм используется); как амплитуда связана с коэффициентом синусоиды (почему умножаем на 2 и делим на длину); как FFT применяется к анализу сезонности успеваемости в школе (выявление периодических пиков); что такое спектр и как его интерпретировать.

````python
import numpy as np

rng = np.random.default_rng(42)
t = np.linspace(0, 1, 1000, endpoint=False)
signal = np.sin(2*np.pi*5*t) + 0.5*np.sin(2*np.pi*12*t)
noisy  = signal + rng.normal(0, 0.3, 1000)

fft   = np.fft.fft(noisy)
freqs = np.fft.fftfreq(len(noisy), d=1/1000)
amp   = np.abs(fft) / len(noisy) * 2

pos = freqs >= 0
top2 = np.argsort(amp[pos])[-2:]
print("топ-2 частоты:", freqs[pos][top2], "амплитуды:", amp[pos][top2])

fft_f = fft.copy()
fft_f[np.abs(freqs) > 20] = 0
filtered = np.fft.ifft(fft_f).real
print("std(noisy-signal)   =", np.std(noisy - signal))
print("std(filtered-signal)=", np.std(filtered - signal))

# FFT раскладывает сигнал на сумму синусоид с частотами.
# Быстрее всего при длине = степени 2 (radix-2 FFT, O(n log n)).
# Амплитуда синусоиды: |X[k]| / N * 2 (множитель 2 — из симметрии).
# Сезонность успеваемости: пики в спектре = периоды (неделя, семестр).
````
