4) Известны площади круга и квадрата. Определить:
а) уместится ли круг в квадрате?
б) уместится ли квадрат в круге?


```
# 4
S_circle = float(input("Введите площадь круга: "))
S_square = float(input("Введите площадь квадрата: "))

# Вычисляем  геометрические параметры
radius = math.sqrt(S_circle / math.pi)  #  радиус круга: R = sqrt(S / pi)
diameter = 2 * radius  #  диаметр круга: D = 2 * R
side = math.sqrt(S_square)  #  сторону квадрата: a = sqrt(S)

# а) Проверяем, поместится ли круг в квадрат
# Круг поместится, если его диаметр меньше или равен стороне квадрата
if diameter <= side:  # Сравниваем диаметр и сторону
    print("а) Круг поместится в квадрате")
else:  # Если диаметр больше стороны
    print("а) Круг не поместится в квадрате")

# б) Проверяем, поместится ли квадрат в круг
# Квадрат поместится, если его диагональ (a * sqrt(2)) меньше или равна диаметру круга
if side * math.sqrt(2) <= diameter:  # Сравниваем диагональ квадрата и диаметр круга
    print("б) Квадрат поместится в круге")
else:  # Если диагональ больше диаметра
    print("б) Квадрат не поместится в круге")
```

а) на поле (a, b) расположена ладья. Определить, угрожает ли она полю
(c,d);
б) на поле (a, b) расположен слон. Определить, угрожает ли он полю (c, d);



```
# 5
print("Введите координаты от 1 до 8:")
a = int(input("Введите a (вертикаль для первой фигуры): "))
b = int(input("Введите b (горизонталь для первой фигуры): "))
c = int(input("Введите c (вертикаль для второй фигуры): "))
d = int(input("Введите d (горизонталь для второй фигуры): "))

# а) Ладья угрожает, если находится на той же вертикали или горизонтали
if a == c or b == d:  # Проверяем совпадение координат
    print("а) Ладья угрожает полю (c, d)")
else:  # Если координаты не совпадают
    print("а) Ладья не угрожает полю (c, d)")

# б) Слон угрожает, если находится на той же диагонали
# Условие диагонали: модуль разности вертикалей равен модулю разности горизонталей
if abs(a - c) == abs(b - d):  # Проверяем условие диагонали
    print("б) Слон угрожает полю (c, d)")
else:  # Если условие не выполняется
    print("б) Слон не угрожает полю (c, d)")
```

1)	Составить программу-генератор простых чисел, в основу положить формулу 2х2+29 при 0≤х≤28.

```
import math

def is_prime(n):  # Объявляем функцию для проверки числа на простоту
    if n <= 1:
        return False  # Оно не является простым, возвращаем False
    for i in range(2, int(math.sqrt(n)) + 1):  # Перебираем делители от 2 до корня из n
        if n % i == 0:  # Если число делится на i без остатка
            return False  # Значит, оно составное, возвращаем False
    return True  # Если делителей не найдено, число простое, возвращаем True

print("Простые числа, полученные по формуле 2x^2 + 29:")
for x in range(0, 29):  # Запускаем цикл, где x принимает значения от 0 до 28 включительно
    value = 2 * (x**2) + 29  # Вычисляем значение по заданной формуле
    if is_prime(value):  # Проверяем, является ли полученное значение простым числом
        print(f"x = {x}, число = {value}")  # Если да, выводим x и полученное простое число

```

```
import math
epsilon = 1e-3  # Задаем точность вычисления (10 в степени -3, то есть 0.001)
total_sum = 0.0  # заводим переменную для накопления суммы ряда
n = 1  # Начинаем с первого члена ряда (n = 1)

while True:  # Запускаем бесконечный цикл для вычисления суммы
    a_n = math.factorial(n) / (3 * (n**n))  # Вычисляем общий член ряда по формуле
    if abs(a_n) < epsilon:  # Если модуль текущего члена ряда меньше заданной точности
        break  # Прерываем цикл, так как достигли нужной точности
    total_sum += a_n  # Прибавляем текущий член ряда к общей сумме
    n += 1  # Увеличиваем n на 1 для перехода к следующему члену

print(f"Сумма ряда с точностью {epsilon} = {total_sum}")
```

1.	Натуральное число, в записи которого n цифр, называется числом Амстронга, если сумма его цифр, возведенная в степень n, равна самому числу. Найти все числа Амстронга от 1 до k.

```
k = int(input("Введите число k (верхнюю границу поиска): "))  #  число k преобразуем  в целое
print(f"Числа Армстронга от 1 до {k}:")

for num in range(1, k + 1):  # Запускаем цикл: num принимает значения от 1 до k включительно
    num_str = str(num)  # Преобразуем текущее число в строку
    n = len(num_str)  # Вычисляем количество цифр в числе (это будет степень n)
    sum_pow = 0  #  переменную для накопления суммы цифр, возведенных в степень n

    for digit_char in num_str:  # Перебираем каждый символ (цифру) в строковом представлении числа
        sum_pow += int(digit_char) ** n  #  символ в целое число, возводим в степень n и добавляем к сумме

    if sum_pow == num:  # Проверяем главное условие: равна ли сумма степеней цифр самому числу
        print(num)  # Если условие выполнено, выводим число на экран
```

2.	Написать функцию вычисления суммы факториалов всех нечетных чисел от 1 до 9.

```
# 2
def sum_odd_factorials():  #  функция для вычисления суммы
    total = 0  #  переменная для хранения итоговой суммы
    for i in range(1, 10):  # Запускаем цикл: i принимает значения от 1 до 9 включительно
        if i % 2 != 0:  # Проверяем, является ли число нечетным (остаток от деления на 2 не равен нулю)
            total += math.factorial(i)  # Вычисляем факториал числа i и прибавляем его к общей сумме
    return total  # Возвращаем полученную сумму из функции

result = sum_odd_factorials()  # Вызываем функцию и сохраняем возвращенное значение в переменную result
print(f"Сумма факториалов нечетных чисел от 1 до 9: {result}")
```

3. Дана символьная строка, представляющая собой запись натурального числа в р-ичной системе счисления (2 ≤ р ≤ 9). Составить программу перевода этого числа в десятичную систему счисления , используя рекурсивную подпрограмму.

```
# 3
def base_p_to_decimal(s, p):  # Объявляем рекурсивную функцию. s - строка с числом, p - основание системы
    if not s:  # Базовый случай рекурсии: если строка пустая (все цифры обработаны)
        return 0  # Возвращаем 0, так как прибавлять больше нечего
    # Рекурсивный шаг:
    # 1. Берем строку без последнего символа (s[:-1]) и переводим её в десятичную.
    # 2. Умножаем результат на основание системы (p).
    # 3. Прибавляем последнюю цифру, преобразованную в целое число (int(s[-1])).
    return base_p_to_decimal(s[:-1], p) * p + int(s[-1])  # Возвращаем вычисленное значение

p = int(input("Введите основание системы счисления p (от 2 до 9): "))  # Запрашиваем основание системы
num_str = input("Введите число в p-ичной системе счисления: ")  # Запрашиваем строку с числом

decimal_number = base_p_to_decimal(num_str, p)  # Вызываем рекурсивную функцию для перевода
print(f"Число {num_str} в {p}-ичной системе = {decimal_number} в десятичной")
```
